```python
# %%capture
from collections import Counter

from tklearn.kb.base import KnowledgeBase

kb = KnowledgeBase("wiktionary")
```

```python
Counter(t.predicate for t in kb.triples)
```

```python
for triple in kb.triples:
    if triple.predicate == "hyponym" and triple.subject[0] == "dog":
        print(triple)
```

```python
kb.triples.get(subject=("dog", 60))
```

```python
list(kb.lexicon.extract("dog is an animal"))
```

```python
kb.lexicon["dog"]
```

```python
for mention in kb.extract_mentions("dog is an animal"):
    for candidate in mention.candidates:
        subject = (candidate.word, candidate.sense_id)
        print(kb.triples.get(subject=subject))
        print(kb.triples.get(subject=subject[0]))
        print(
            mention.form,
            candidate.word,
            candidate.sense_id,
            candidate.definition,
        )
```
