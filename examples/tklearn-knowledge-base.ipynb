```python
from tklearn.kb.base import KnowledgeBase

kb = KnowledgeBase("wiktionary")
```

```python
for key, value in kb.gloss2idx.items():
    print(f"{key}: {value}")
    break

kb.embeddings[0].shape
```

```python
word = "coconut"

for sense_id in kb.senses[word]:
    subject = (word, sense_id)
    print(sense_id, kb.idx2gloss[sense_id])
    print("=>", kb.triples.get(subject))
    print()
```

```python
example_text = "that coconut is the most racist fucking idiot i have ever met"
```

```python
for mention in kb.extract_mentions(example_text):
    for candidate in mention.candidates:
        if candidate.sense_id is None:
            assert candidate.definition is None
            continue
        for subject in [
            (candidate.word, None),
            (candidate.word, candidate.sense_id),
        ]:
            sense_triples = kb.triples.get(subject, {})
            for predicate, objects in sense_triples.items():
                print(
                    subject + (predicate,),
                    "=>",
                    objects,
                    candidate.definition,
                )
                print()
```
