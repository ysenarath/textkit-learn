```python
from tklearn.embeddings import AutoEmbedding
```

```python
wv = AutoEmbedding("glove-twitter-25")

assert wv["hello"].shape == (25,)
```

```python
wv = AutoEmbedding({
    "loader": "gensim",
    "name": "glove-twitter-25",
})

assert wv["hello"].shape == (25,)
```

```python
wv = AutoEmbedding({
    "loader": "fasttext",
    "name": "cc.en.300.bin",
})

assert wv["hello"].shape == (300,)
```
